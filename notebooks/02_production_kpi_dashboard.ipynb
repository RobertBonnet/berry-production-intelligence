# Production KPI Dashboard

## Business Objective

Analyze berry farm production data to generate insights.

```python

import pandas as pd

data = pd.read_csv("../data/farm_production_data.csv")

data["yield_efficiency"] = data["actual_yield_kg"] / data["expected_yield_kg"]

kpis = {
"avg_yield_efficiency": data["yield_efficiency"].mean(),
"total_production": data["actual_yield_kg"].sum()
}

kpis
```
